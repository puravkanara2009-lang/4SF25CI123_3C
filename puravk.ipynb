```python
# Safely drop 'Unnamed: 0' only if it exists in the DataFrame
if 'Unnamed: 0' in df.columns:
    df = df.drop(columns=['Unnamed: 0'])
    print("Dropped 'Unnamed: 0' successfully.")
else:
    print("Column 'Unnamed: 0' is not present in this dataset. Skipping...")

print(df.head())
```

```python
# changing datatype
df['case_id']=df['case_id'].astype(float)
df['case_id'].head()
```

```python
# adding new columns
df['case_year'] = df['date'].dt.year
df['summary_length'] = df['summary'].str.len()
df['is_granted'] = (df['bail_outcome'] == 'Granted').astype(int)
print('New columns added successfully!')
print(df[['case_id', 'case_year', 'summary_length', 'is_granted']].head())
```

```python
# checking null values
df.isnull().sum()
```

```python
# filling missing text values and numeric derived values
text_columns = df.select_dtypes(include=['object']).columns
for column in text_columns:
    df[column] = df[column].fillna('Unknown')
df['summary_length'] = df['summary_length'].fillna(df['summary_length'].mean())
```

```python
# checking null values after filling
df.isnull().sum()
```

```python
df['double_summary_length'] = df['summary_length'].apply(lambda x: x * 2)
print(df[['summary_length', 'double_summary_length']].head())
```

```python
# grouping accused gender and finding the bail-grant rate
gender_outcome = df.groupby('accused_gender')['is_granted'].mean().sort_values(ascending=False)
print(gender_outcome)
```

```python
# grouping crime type and finding the average summary length
crime_summary = df.groupby('crime_type')['summary_length'].mean().sort_values(ascending=False)
print(crime_summary)
```

```python
# grouping court and finding the bail-grant rate
court_outcome = df.groupby('court')['is_granted'].mean().sort_values(ascending=False)
print(court_outcome.head(10))
```

```python
# grouping region and finding the number of cases
region_counts = df.groupby('region')['case_id'].count().sort_values(ascending=False)
print(region_counts.head(10))
```

```python
# counting cases for each crime type
crime_counts = df.groupby('crime_type')['case_id'].count()
print(crime_counts)
```

```python
# sorting number of cases per crime type in descending order
crime_counts = df.groupby('crime_type')['case_id'].count().sort_values(ascending=False)
print(crime_counts.head(10))
```

```python
# grouping using bail outcome and finding average case indicators
outcome_analysis = df.groupby('bail_outcome')[['summary_length', 'is_granted', 'bail_cancellation_case', 'landmark_case']].mean()
print(outcome_analysis)
```

```python
# sorting cases by summary length in descending order
longest_summaries = df.sort_values('summary_length', ascending=False)
print(longest_summaries[['case_id', 'court', 'summary_length']].head(10))
```

```python
# sorting cases by date in descending order
latest_cases = df.sort_values('date', ascending=False)
print(latest_cases[['case_id', 'date', 'bail_outcome']].head(10))
```

```python
# counting cases for each court
court_counts = df.groupby('court')['case_id'].count()
top_courts = court_counts.sort_values(ascending=False).head(10)
print(top_courts)
```

```python
# counting cases for each region
region_counts = df.groupby('region')['case_id'].count()
top_regions = region_counts.sort_values(ascending=False).head(10)
print(top_regions)
```

```python
# bar graph for top 10 regions by case count
top_regions.plot(kind='bar')
plt.xlabel('Region', fontsize=12)
plt.ylabel('Number of Cases', fontsize=12)
plt.title('Top 10 Regions by Number of Bail Cases', fontsize=14, fontweight='bold')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
# bar graph for bail outcomes
outcome_counts = df['bail_outcome'].value_counts()
outcome_counts.plot(kind='bar')
plt.xlabel('Bail Outcome')
plt.ylabel('Number of Cases')
plt.title('Bail Outcome Counts')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# scatter plot for case year vs summary length
plt.scatter(df['case_year'], df['summary_length'])
plt.xlabel('Case Year')
plt.ylabel('Summary Length')
plt.title('Case Year vs Summary Length')
plt.show()
```

```python
# scatter plot for landmark cases vs bail outcome
plt.scatter(df['landmark_case'].astype(int), df['is_granted'])
plt.xlabel('Landmark Case (0 = No, 1 = Yes)')
plt.ylabel('Granted Outcome (0 = No, 1 = Yes)')
plt.title('Landmark Cases vs Bail Outcome')
plt.show()
```

```python
# grouping bail outcome and finding average summary length
outcome_summary = df.groupby('bail_outcome')['summary_length'].mean()
print(outcome_summary)
```

```python
# histogram for case years
plt.hist(df['case_year'], bins=10)
plt.xlabel('Case Year')
plt.ylabel('Number of Cases')
plt.title('Distribution of Case Years')
plt.show()
```

```python
# histogram for summary lengths
plt.hist(df['summary_length'], bins=10)
plt.xlabel('Summary Length')
plt.ylabel('Number of Cases')
plt.title('Distribution of Summary Lengths')
plt.show()
```

```python
# pie chart for bail outcome distribution
outcome_count = df['bail_outcome'].value_counts()
plt.pie(outcome_count, labels=outcome_count.index, autopct='%1.1f%%')
plt.title('Bail Outcome Distribution')
plt.show()
```
